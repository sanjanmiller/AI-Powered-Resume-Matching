```python
import pymupdf
def extract_text_from_pdf(path):
    doc = pymupdf.open(path)

    pages = []

    for page in doc:
        pages.append(page.get_text())

    doc.close()

    return "\n".join(pages)
```

```python
import re

def clean_text(text):
    # Remove excessive whitespace while preserving line structure
    lines = [line.strip() for line in text.splitlines()]

    # Remove empty lines
    lines = [line for line in lines if line]

    # Remove standalone page numbers
    lines = [line for line in lines if not re.fullmatch(r"\d+", line)]

    return "\n".join(lines)
```

```python
cleaned_text = clean_text(resume_texts[first_pdf])

print(cleaned_text)
```

```python
sections = [
    "Professional Summary",
    "Technical Skills",
    "Work Experience",
    "Projects",
    "Education",
    "Certifications"
]
```

```python
def section_aware_chunking(text):
    lines = text.splitlines()

    chunks = []
    current_section = None
    current_content = []

    for line in lines:
        # Check whether this line is a section heading
        if line in sections:

            # Save the previous section
            if current_section and current_content:
                chunks.append({
                    "section": current_section,
                    "text": "\n".join(current_content).strip()
                })

            # Start the new section
            current_section = line
            current_content = []

        else:
            # Add content to the current section
            if current_section:
                current_content.append(line)

    # Save the final section
    if current_section and current_content:
        chunks.append({
            "section": current_section,
            "text": "\n".join(current_content).strip()
        })

    return chunks
```

```python
chunks = section_aware_chunking(cleaned_text)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print("Section:", chunk["section"])
    print(chunk["text"])
```

```python
all_chunks = []

for pdf in pdf_files:
    # Get candidate text
    text = resume_texts[pdf]

    # Clean text
    text = clean_text(text)

    # Create section-aware chunks
    chunks = section_aware_chunking(text)

    # Add candidate information
    for chunk in chunks:
        all_chunks.append({
            "candidate": pdf.replace(".pdf", ""),
            "source_file": pdf,
            "section": chunk["section"],
            "text": f"Section: {chunk['section']}\n\n{chunk['text']}"
        })

print("Total chunks:", len(all_chunks))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
chunk_texts = [
    chunk["text"]
    for chunk in all_chunks
]

print("Number of texts:", len(chunk_texts))
```

```python
embeddings = embedding_model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Embedding shape:", embeddings.shape)

#26 → number of resume chunks
#384 → embedding dimensions produced by MiniLM
```

```python
import chromadb

client = chromadb.PersistentClient(
    path="./chroma_db"
)
```

```python
collection = client.get_or_create_collection(
    name="resume_knowledge_base"
)

print("Collection created:", collection.name)
```

```python
ids = [
    f"{chunk['candidate']}_{i}"
    for i, chunk in enumerate(all_chunks)
]

metadatas = [
    {
        "candidate": chunk["candidate"],
        "source_file": chunk["source_file"],
        "section": chunk["section"]
    }
    for chunk in all_chunks
]
```

```python
collection.add(
    ids=ids,
    documents=chunk_texts,
    embeddings=embeddings.tolist(),
    metadatas=metadatas
)

print("Chunks stored:", collection.count())
```

```python
result = collection.get(
    limit=3,
    include=["documents", "metadatas"]
)

for i in range(len(result["ids"])):
    print("\nID:", result["ids"][i])
    print("Metadata:", result["metadatas"][i])
    print("Document:")
    print(result["documents"][i][:300])
```

```python
sample_jd = """
We are looking for a Machine Learning Engineer with experience developing
and deploying machine learning models.

The candidate should have strong Python and Scikit-learn skills,
experience with PyTorch, FastAPI and REST APIs, and experience
containerizing applications using Docker.

Experience deploying machine learning workloads on AWS is preferred.
Knowledge of data preprocessing and feature engineering is also required.
"""
```

```python
jd_embedding = embedding_model.encode(
    sample_jd
)

print("JD embedding shape:", jd_embedding.shape)
```

```python
results = collection.query(
    query_embeddings=[jd_embedding.tolist()],
    n_results=10
)
```

```python
# ChromaDB returns a distance between the JD embedding and each resume chunk.
# Lower distance = greater semantic similarity.
# This is a retrieval signal, NOT a candidate hiring/suitability score.

for i in range(len(results["documents"][0])):
    print(f"\n{'=' * 60}")
    print(f"Result {i + 1}")
    print("Candidate:", results["metadatas"][0][i]["candidate"])
    print("Section:", results["metadatas"][0][i]["section"])
    print("Distance:", results["distances"][0][i])
    print("-" * 60)
    print(results["documents"][0][i][:500])
```

```python
from collections import defaultdict

# Group retrieved resume chunks by candidate.
# A candidate may appear multiple times because multiple sections
# of their resume can be relevant to the JD.

candidate_results = defaultdict(list)

for i in range(len(results["documents"][0])):

    candidate = results["metadatas"][0][i]["candidate"]

    candidate_results[candidate].append({
        "section": results["metadatas"][0][i]["section"],
        "document": results["documents"][0][i],
        "distance": results["distances"][0][i]
    })


# Display the grouped results.
for candidate, chunks in candidate_results.items():

    print(f"\n{'=' * 60}")
    print("Candidate:", candidate)
    print("Retrieved chunks:", len(chunks))

    for chunk in chunks:

        # Lower semantic retrieval distance means
        # greater semantic similarity to the JD.
        print(
            f"  - {chunk['section']} "
            f"(semantic retrieval distance: {chunk['distance']:.3f})"
        )
```

```python
for candidate, chunks in candidate_results.items():

    print(f"\n{'=' * 70}")
    print(f"Candidate: {candidate}")

    for chunk in chunks:
        print(f"\n--- {chunk['section']} ---")
        print(f"Semantic retrieval distance: {chunk['distance']:.3f}")
        print(chunk["document"])
```
